```python
def check_sizes(df, ts_path):
    max_size = 0
    min_size = 100000

    for file_name in df.ism330dhcx_acc:
        file_path = os.path.join(ts_path, file_name)
        ts = pd.read_parquet(file_path)
        size = len(ts)
        if size > max_size:
            max_size = size
        if size < min_size:
            min_size = size

    return max_size, min_size


ts_path = os.path.join(TS_PATH, "test")

df_anomaly_source_max_size, df_anomaly_source_min_size = check_sizes(
    df_anomaly_source, ts_path
)
df_anomaly_target_max_size, df_anomaly_target_min_size = check_sizes(
    df_anomaly_target, ts_path
)
df_normal_source_max_size, df_normal_source_min_size = check_sizes(
    df_normal_source, ts_path
)
df_normal_target_max_size, df_normal_target_min_size = check_sizes(
    df_normal_target, ts_path
)

print(
    "min_size =", df_anomaly_source_min_size, "  max_size =", df_anomaly_source_max_size
)
print(
    "min_size =", df_anomaly_target_min_size, "  max_size =", df_anomaly_target_max_size
)
print(
    "min_size =", df_normal_source_min_size, "  max_size =", df_normal_source_max_size
)
print(
    "min_size =", df_normal_target_min_size, "  max_size =", df_normal_target_max_size
)
```

```python
def get_plot(ts, file_path_save, title):

    fig, axes = plt.subplots(nrows=3, ncols=1, figsize=(14, 8), sharex=True)

    axes[0].plot(ts.index, ts["A_x [g]"], color="#1f77b4", linewidth=0.5)
    axes[0].set_ylabel("A_x [g]", fontweight="bold")
    axes[0].grid(True, linestyle="--", alpha=0.5)
    # axes[0].set_title(title, fontsize=12)

    axes[1].plot(ts.index, ts["A_y [g]"], color="#ff7f0e", linewidth=0.5)
    axes[1].set_ylabel("A_y [g]", fontweight="bold")
    axes[1].grid(True, linestyle="--", alpha=0.5)

    axes[2].plot(ts.index, ts["A_z [g]"], color="#2ca02c", linewidth=0.5)
    axes[2].set_ylabel("A_z [g]", fontweight="bold")
    axes[2].set_xlabel("Time step", fontweight="bold")
    axes[2].grid(True, linestyle="--", alpha=0.5)

    plt.tight_layout()

    plt.savefig(file_path_save, format="pdf")
    plt.show()
```

```python
import numpy as np
import pandas as pd
from scipy.stats import kurtosis, skew

comb = (
    df[["anomaly_label", "domain_shift_op"]]
    .drop_duplicates()
    .reset_index(drop=True)
    .values
)

results = []

for anomaly_label, domain_shift_op in comb:

    act = all_data[
        (all_data["anomaly_label"] == anomaly_label)
        & (all_data["domain_shift_op"] == domain_shift_op)
    ].reset_index(drop=True)

    for j in range(len(act)):
        file_name = act.loc[j, "ism330dhcx_acc"]
        split_label = act.loc[j, "split_label"]

        if str(split_label).endswith("Train"):
            folder_path = "train"
        else:
            folder_path = "test"

        ts = pd.read_parquet(
            f"../data/03_primary/time_series_data/{folder_path}/{file_name}"
        ).values

        ts_x = ts[1:, 1]
        ts_y = ts[1:, 2]
        ts_z = ts[1:, 3]

        kurt_x = kurtosis(ts_x, fisher=False)
        kurt_y = kurtosis(ts_y, fisher=False)
        kurt_z = kurtosis(ts_z, fisher=False)

        mean_x = np.mean(ts_x)
        mean_y = np.mean(ts_y)
        mean_z = np.mean(ts_z)

        std_x = np.std(ts_x)
        std_y = np.std(ts_y)
        std_z = np.std(ts_z)

        skew_x = skew(ts_x)
        skew_y = skew(ts_y)
        skew_z = skew(ts_z)

        results.append(
            {
                "anomaly_label": anomaly_label,
                "domain_shift_op": domain_shift_op,
                "file_name": file_name,
                "kurt_x": kurt_x,
                "kurt_y": kurt_y,
                "kurt_z": kurt_z,
                "mean_x": mean_x,
                "mean_y": mean_y,
                "mean_z": mean_z,
                "std_x": std_x,
                "std_y": std_y,
                "std_z": std_z,
                "skew_x": skew_x,
                "skew_y": skew_y,
                "skew_z": skew_z,
            }
        )

results_df = pd.DataFrame(results)
results_df.to_csv("../data/08_reporting/eda_statistics.csv")
```

```python
metric_cols = [
    "kurt_x",
    "kurt_y",
    "kurt_z",
    "mean_x",
    "mean_y",
    "mean_z",
    "std_x",
    "std_y",
    "std_z",
    "skew_x",
    "skew_y",
    "skew_z",
]

summary_df = (
    results_df.groupby(["anomaly_label", "domain_shift_op"])[metric_cols]
    .agg(["mean", "std"])
    .reset_index()
)

summary_df.columns = [
    f"{col[0]}_{col[1]}" if col[1] else col[0] for col in summary_df.columns
]

summary_df.to_csv("../data/08_reporting/eda_metrics_summary.csv")
```
